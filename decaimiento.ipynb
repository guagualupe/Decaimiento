```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import warnings
warnings.filterwarnings("ignore")

# Función de decaimiento
def N(t):
    return 8 * (1/2)**(t / 25)

# Gráfico interactivo con deslizador
@interact(t=(0, 150, 1))
def plot_graph(t=0):
    x = np.linspace(0, 150, 300)
    y = N(x)
    punto = N(t)

    plt.figure(figsize=(10, 5))
    plt.plot(x, y, label='N(t) = 8*(1/2)^(t/25)')
    plt.axhline(0.4, color='r', linestyle='--', label='y = 0.4')  # Línea roja en 0.4
    plt.axvline(t, color='g', linestyle=':', label=f't = {t}')    # Línea vertical móvil
    plt.scatter([t], [punto], color='black')                      # Punto móvil
    plt.xlabel('t (años)')
    plt.ylabel('N(t) (gramos)')
    plt.ylim(0, 9)
    plt.grid(True)
    plt.legend()
    plt.title('Función de decaimiento con línea y = 0.4')
    plt.show()
```
