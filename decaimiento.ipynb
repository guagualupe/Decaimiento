```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import warnings
warnings.filterwarnings("ignore")

# Función exponencial corregida: ahora acepta el parámetro t
def N(t):
    return 8 * (1/2)**(t / 25)

# Función interactiva que grafica con el valor de t
@interact(t=(0, 150, 1))
def plot_graph(t=0):
    x = np.linspace(0, 150, 300)
    y = 8 * (1/2)**(x / 25)  # toda la curva
    punto = N(t)  # valor puntual

    plt.figure(figsize=(10, 5))
    plt.plot(x, y, label='N(t) = 8*(1/2)^(t/25)')
    plt.axhline(1, color='r', linestyle='--', label='y = 1')
    plt.axvline(t, color='g', linestyle=':', label=f't = {t}')
    plt.scatter([t], [punto], color='black')
    plt.xlabel('t (años)')
    plt.ylabel('N(t) (gramos)')
    plt.ylim(0, 9)
    plt.grid(True)
    plt.legend()
    plt.title('Función de decaimiento')
    plt.show()
```
